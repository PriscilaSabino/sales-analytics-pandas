```python
#Importação / Visualização

import pandas as pd
import numpy as np

df_vendas = pd.read_csv("../data/raw/vendas_tech.csv", low_memory=False)
df_gerentes = pd.read_excel("../data/raw/gerentes_lojas.xlsx")

display(df_vendas.info())
display(df_gerentes.info())

```

```python
# Limpeza da tabela vendas

# 1. Deleta a coluna irrelevante
#df_vendas = df_vendas.drop(columns=["Data_Base"])

# 2. Converte a coluna Data para datetime
df_vendas["Data"] = pd.to_datetime(df_vendas["Data"], format="%Y-%m-%d")

# # 3. Limpa espaços extras no texto das lojas
df_vendas["Loja"] = df_vendas["Loja"].str.strip()

# # 4. Preenche os valores ausentes na coluna Loja com "Online"
df_vendas["Loja"] = df_vendas["Loja"].fillna("Online").str.title()

# # 5. Remove pedidos duplicados
df_vendas = df_vendas.drop_duplicates(subset=["ID_Pedido", "Loja"])

# # Limpeza da tabela gerente

df_gerentes["Loja"] = df_gerentes["Loja"].str.title()

# # Unindo tabelas
df_analise = df_vendas.merge(df_gerentes, left_on="Loja", right_on="Loja", how="left")

# Preencher nulos resultantes do merge
df_analise["Gerente"] = df_analise["Gerente"].fillna("Sem Gerente (E-commerce)")
df_analise["Meta_Mensal"] = df_analise["Meta_Mensal"].fillna(0)
```

```python
# Criando novas tabelas
df_analise["Faturamento"] = df_analise["Qtd"] * df_analise["Preco_Unitario"]
df_analise['Ano_Mes'] = df_analise['Data'].dt.to_period('M')
display(df_analise)
df_analise.to_csv("vendas_tratadas.csv")
```
